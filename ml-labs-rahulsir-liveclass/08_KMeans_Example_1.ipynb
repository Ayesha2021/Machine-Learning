#make_blobs generate random data

n_features=2 400 rows of data has two columns
centers=4 it groups 400 row 2 columns data  in to four groups.
cluster_std=1.2 : mean distance of each datapoint from center of their cluster
random_state=42 means me and rahul sir get same set of data from training.

```python
data =make_blobs(n_samples=400,n_features=2,centers=4,cluster_std=1.2,random_state=42)

type(data)
```

Unsupervised Learning

```python
kmeans = KMeans(n_clusters =4)
kmeans
```

```python
kmeans.fit(data[0])
```

```python
labels = kmeans.labels_
labels
```

plot with one row two columns consisting for one columsn plot with normal scatter plot other with kmeans fit data plotting --how alogrithm learned patterns from data.

```python
fig, (ax1,ax2) = plt.subplots(1, 2, figsize=(10,6))  # unpacking 
ax1.set_title('Origional plot using matplot lib')
ax1.scatter(data[0][:,0],data[0][:,1],cmap='rainbow', c=data[1])
ax2.set_title('K-Means Pattern Generation')
ax2.scatter(data[0][:,0],data[0][:,1], c=kmeans.labels_,cmap='rainbow')
```

Silhoutte Score https://scikit-learn.org/stable/modules/clustering.html#silhouette-coefficient

Silhoutte Score Range == -1 to 1
Interpret the Score
+1 : Indicates that the datapoints are far from the neighboring clusters  and well   matched to its own clusters.
0 : Suggest to Data points is very close to the decision boundry between two clusters.
-1 : Means the data point has been assigned to wrong cluster.

```python
sil_score = silhouette_score(data[0],labels)
print('Slhautte Score :',sil_score)
```
