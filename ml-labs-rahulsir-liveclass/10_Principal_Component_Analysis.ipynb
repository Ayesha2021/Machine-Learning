```python
import pandas as pd 
from sklearn.datasets import load_digits
from sklearn.decomposition import PCA
%matplotlib inline
```

```python
#load digit dataset
digits =load_digits()
type(digits)
```

```python
digits.keys() #digit is like a dictionary
```

```python
print(digits['DESCR'])
```

```python
digits.feature_names  # or you can use digits['feature_names']
```

```python
digits.target  #digits made of 0 to 9 it is also the target
```

```python
digits.data
```

```python
X= digits.data
X
```

```python
#Reduce from 64  dimension to 2 dimentions
pca = PCA(n_components=2)   #n_components is the number of dimensions
pca.fit(X)
X=pca.transform(X)
X
```

Above array consist of principle componets 1 and 2---1797 are the number of digits...attributes reduce from 64 to 2

```python
#Components of the principal components
pca.components_
```

```python
pca.components_.shape
```

```python
df_components =pd.DataFrame(pca.components_)
df_components
```

```python
df_components =pd.DataFrame(pca.components_,columns=digits.feature_names)
df_components
```

```python
df_components.shape
```

```python
#plot the transformed data
plt.scatter(X[:,0], X[:,1], c=digits.target,cmap='tab10',s=10)
plt.xlabel('First Principal Component 1')
plt.ylabel('Second Principal Component 2')
plt.colorbar()
plt.show()
```
