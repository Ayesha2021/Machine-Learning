```python
### Measure of Center Mean Median Mode range
mean = titanic['age'].mean()
median = titanic['age'].median()
mode = titanic['age'].mode()

print(f'mean = {mean}')
print(f'median = {median}')
print(f'mode = {mode[0]}')
```

```python
### Measure of Center for fares
```

```python
meanFare = titanic['fare'].mean()
medianFare =titanic['fare'].median()
modeFare = titanic['fare'].mode()
print(f'Mean: {meanFare}')
print(f'Median: {medianFare}')
print(f'Mode: {modeFare[0]}')
```

```python
### Observation 
### Mean fare> Median fare means -----Mean is far from median fare,(Right skewed)
### in case of age Mean and mdian are close,which implies most of the passengers are around late20
```

```python
### Measure of Spread - Standard Deviation ,variance ,quartiles,range  for age
```

```python
minAge = titanic['age'].min()
maxAge = titanic['age'].max()
print(f'min and max of age are ',{minAge}, {maxAge})
std =titanic['age'].std()
print(f'std of age is {std}')
variance =titanic['age'].var()
print(f'variance of age is {variance}')
q1 =titanic['age'].quantile(0.25)
q3 = titanic['age'].quantile(0.3)
iqr = q3 - q1
print(f'inter quartile range of age is {iqr}')
```

```python
###Measure of spread for fare
std=titanic['fare'].std()
print("Standard Deviation of Fare is",std)
var=titanic['fare'].var()
print("Variance of Fare is",var)
minFare =titanic['fare'].min()
print("Minimum Fare is",minFare)
maxFare =titanic['fare'].max()
print("Maximum Fare is",maxFare)
q1=titanic['fare'].quantile(0.25)
q3=titanic['fare'].quantile(0.75)
iqr=q3-q1
print("Inter Quartile Range of Fare is",iqr)
```

```python
###Shape of Distribution 
print('Skew Age:',titanic['age'].skew())
print('Kurtosis Age:',titanic['age'].kurt())
```

```python
print('Fare Skew ::', titanic['fare'].skew())
print('Fare Kurtosis::', titanic['fare'].kurtosis())
```
