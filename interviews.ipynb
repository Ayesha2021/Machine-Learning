```python
import pandas as pd
import numpy as np
```

```python
df1= pd.DataFrame(
    {'A':['A0','A1']}, index=['K0','K1']
)

df2 =pd.DataFrame(
    {'B': ['B0','B1']}, index=['K0','K2']
)

joined = df1.join(df2) # default is left join and join based on dataframe key

joined
```

```python
# df.concat

df3 =pd.DataFrame(
    {'item':['book','pen'],'quantiy':[10,20]}
)
df4 =pd.DataFrame(
    {'item':['Notebook','eraso'],'quantiy':[30,40]}
)

total_sales = pd.concat([df3,df4],axis=0,ignore_index=True)

print(total_sales)
total_sales1 = pd.concat([df3,df4],axis=1,ignore_index=True)

print(total_sales1)
```
