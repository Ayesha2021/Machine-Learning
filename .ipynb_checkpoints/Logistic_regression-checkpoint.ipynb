```python
import numpy as np
from matplotlib import pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
import pandas as pd

```

```python
df =pd.read_csv("data/diabetes.csv")
print(len(df))
```

```python
#separate Feature
df.columns
```

```python
x,y =  df[['PatientID', 'Pregnancies', 'PlasmaGlucose', 'DiastolicBloodPressure',
       'SerumInsulin', 'BMI', 'DiabetesPedigree', 'Age',]].values,df['Diabetic'].values
```

```python
#split data in to train and test

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)

len(x_train)
```

```python
diabetic_count =df['Diabetic'].value_counts()
print(diabetic_count)
print(f'Diabetic count is {diabetic_count[0]}')

#modl =LinearRegression()
```

```python
fig =plt.figure(figsize=(10,10))
ax =fig.gca()


ax.set_xlabel('Diabetic')
ax.set_ylabel('No of Paitients')
diabetic_count.plot(kind='bar')
plt.show()    
```

```python
reg =0.01
# C is regularization strength

model = LogisticRegression(C=1/reg,max_iter=500).fit(x_train,y_train)

h_hat = model.predict(x_test)

accuracy =np.average(h_hat == y_test)

print(f'Accuracy is {accuracy:.4f}')

y_pred = model.predict(x_test)
accuracy = accuracy_score(y_test, h_hat)
print("Accuracy:", accuracy)
```
